```python
import requests
from bs4 import BeautifulSoup

def scrape_product(url):
        headers = {
            'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
        }
        response = requests.get(url, headers=headers)
        soup = BeautifulSoup(response.content, 'html.parser')
        pack = soup.findAll(class_= 'bama-vehicle-detail-with-icon__detail-holder')
        try:
            Price = int(soup.find(class_= 'bama-ad-detail-price__price-text').text.replace(',',''))
        except:
            Price = soup.find(class_= 'bama-ad-detail-price__price-text').text
        try:
            Mileage = int(pack[0].p.text.replace('km','').replace(',',''))
        except:
            Mileage = pack[0].p.text
            
        Color = pack[4].p.text
        Production_year = int(soup.find(class_= 'bama-ad-detail-title__subtitle').text.strip())
        Transmission_type = pack[2].p.text
        try:
            Description = soup.find(class_ = 'desc').text
        except:
            Description = ''
            
        return Price, Mileage, Color, Production_year, Transmission_type, Description
    
    

main_page = open('./data.html').read()  
main_soup = BeautifulSoup(main_page)
product_links = [x['href'] for x in main_soup.findAll(class_ = 'bama-ad listing')]
data = [scrape_product(url) for url in product_links]
```

```python
import pandas as pd

df = pd.DataFrame(data,columns=['Price', 'Mileage', 'Color', 'Production_year', 'Transmission_type', 'Description'])
df.to_excel('bama..xlsx')
```
